```python
from math import sin, cos, sqrt, atan2, radians

def calculate_distance(lat1, lon1, lat2, lon2):
    # approximate radius of earth in km
    R = 6373.0

    lat1 = radians(lat1)
    lon1 = radians(lon1)
    lat2 = radians(lat2)
    lon2 = radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2)**2 + cos(lat1) * cos(lat2) * sin(dlon / 2)**2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    distance = R * c
    return distance
```

```python
# Coordenada del launch site
launch_site_coord = [28.562302, -80.577356]

# Coordenadas ficticias de los puntos cercanos
closest_city_coord = [28.61000, -80.60000]
closest_railway_coord = [28.57000, -80.58000]
closest_highway_coord = [28.55000, -80.56000]
```

```python
from math import radians, cos, sin, asin, sqrt

def calculate_distance(lat1, lon1, lat2, lon2):
    """Distancia entre dos puntos geográficos en km (fórmula de Haversine)"""
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = sin(dlat/2)**2 + cos(lat1) * cos(lat2) * sin(dlon/2)**2
    c = 2 * asin(sqrt(a))
    r = 6371
    return c * r
```

```python
from folium.features import DivIcon

# Lista de puntos con nombre y coordenadas
points_of_interest = {
    'Closest City': closest_city_coord,
    'Closest Railway': closest_railway_coord,
    'Closest Highway': closest_highway_coord
}

# Añadir cada marcador + línea al mapa
for name, coord in points_of_interest.items():
    # Calcular distancia
    dist = calculate_distance(launch_site_coord[0], launch_site_coord[1], coord[0], coord[1])
    
    # Crear marcador con texto
    marker = folium.Marker(
        location=coord,
        icon=DivIcon(
            icon_size=(20, 20),
            icon_anchor=(0, 0),
            html=f'<div style="font-size: 12px; color:#0000FF;"><b>{name}<br>{dist:.2f} KM</b></div>'
        )
    )
    
    # Añadir marcador al mapa
    site_map.add_child(marker)
    
    # Dibujar línea al launch site
    folium.PolyLine(
        locations=[launch_site_coord, coord],
        color='green',
        weight=2,
        dash_array='5,5'
    ).add_to(site_map)
```
